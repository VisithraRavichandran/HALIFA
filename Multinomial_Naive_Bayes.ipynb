```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
```

```python
# 1. Prepare simulated production training data
training_data = {
    'text': [
        "Congratulations! You've won a free ticket to the lottery prize draw!",
        "Urgent: Claim your secret bonus rewards account login details now.",
        "Winner of the cash prize click this link immediately to claim.",
        "Hey friend, are we still meeting up for lunch tomorrow at noon?",
        "Can you send over the review notes for the team project presentation?",
        "Hello, just wanted to check if you received my previous email about the schedule."
    ],
    'label': ['spam', 'spam', 'spam', 'ham', 'ham', 'ham']
}

df = pd.DataFrame(training_data)
df
```

```python
from google.colab import sheets
sheet = sheets.InteractiveSheet(df=df)
```

```python
# 2. Build an end-to-end production ML pipeline
# CountVectorizer handles word breaking & count tracking.
# MultinomialNB defaults to alpha=1.0 for Laplace Smoothing.
pipeline = make_pipeline(
    CountVectorizer(),
    MultinomialNB(alpha=1.0)
)
```

```python
# 3. Train the model parameters
pipeline.fit(df['text'], df['label'])
```

```python
# 4. Simulate real-time streaming incoming queries
incoming_emails = [
    "Secret prize winner entry available!",
    "Hey friend, let's meet tomorrow." ,
    "You won 10 Millon dollars !, tell me your bank account number" ,
    "I want to go for shopping"
]
```

```python
# 5. Execute live inference predictions
predictions = pipeline.predict(incoming_emails)
probabilities = pipeline.predict_proba(incoming_emails)
predictions
```

```python
# 6. Output live decision findings
print("--- Real-Time Classification Engine ---")
for i, email in enumerate(incoming_emails):
    print(f"\nIncoming Input: '{email}'")
    print(f"Predicted Tag : {predictions[i].upper()}")
    print(f"Confidence     : [Ham: {probabilities[i][0]:.2%}, Spam: {probabilities[i][1]:.2%}]")
```

```python
import joblib
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import make_pipeline
```

```python
# 1. Dummy production training set
data = {
    'text': ["won a free lottery prize", "secret bonus rewards claim", "lunch tomorrow at noon", "project review presentation"],
    'label': ['spam', 'spam', 'ham', 'ham']
}
df = pd.DataFrame(data)
df
```

```python
# 2. Fit pipeline components
production_pipeline = make_pipeline(CountVectorizer(), MultinomialNB(alpha=1.0))
production_pipeline.fit(df['text'], df['label'])
```

```python
# 3. Export model structure locally
joblib.dump(production_pipeline, 'naive_bayes_spam_model.pkl')
print("Model pipeline successfully serialized to disk!")
```

```python
import joblib

# 1. Load the operational asset back into memory
live_pipeline = joblib.load('naive_bayes_spam_model.pkl')

def predict_stream(incoming_payload: str):
    """
    Simulates parsing text from a live network request event.
    """
    prediction = live_pipeline.predict([incoming_payload])[0]
    probabilities = live_pipeline.predict_proba([incoming_payload])[0]

    # Track metrics matching class labels mapping order
    classes = live_pipeline.classes_
    prob_dict = {classes[i]: f"{probabilities[i]:.2%}" for i in range(len(classes))}

    return {
        "status": "success",
        "input_text": incoming_payload,
        "prediction": prediction.upper(),
        "confidence_scores": prob_dict
    }

# 2. Test dynamic lookup calls
print(predict_stream("Urgent project review bonus needed tomorrow!"))
```

```python
import email
from email.header import decode_header
import imaplib
import joblib

# 1. Configuration Settings
IMAP_SERVER = "outlook.office365.com"  # Use "outlook.office365.com" for Outlook
EMAIL_ACCOUNT = "visithra@halifa-bobo.com"
APP_PASSWORD = "abcd efgh ijkl mnop"  # Your 16-character App Password

# 2. Load your pre-trained Naive Bayes pipeline
try:
    live_pipeline = joblib.load("naive_bayes_spam_model.pkl")
    print("✓ Machine Learning model loaded successfully.")
except FileNotFoundError:
    print(
        "Error: 'naive_bayes_spam_model.pkl' not found. Run your training script first!"
    )
    exit()


def clean_text_header(header_value):
    """Decodes email headers safely."""
    decode_bytes, encoding = decode_header(header_value)[0]
    if isinstance(decode_bytes, bytes):
        return decode_bytes.decode(encoding or "utf-8", errors="ignore")
    return decode_bytes


def fetch_and_test_live_emails():
    print(f"Connecting to {IMAP_SERVER}...")

    # Connect to the server securely over SSL
    mail = imaplib.IMAP4_SSL(IMAP_SERVER)
    mail.login(EMAIL_ACCOUNT, APP_PASSWORD)

    # Select the inbox folder (readonly=True means we won't accidentally delete anything)
    mail.select("inbox", readonly=True)

    # Search for all UNREAD messages
    status, response_data = mail.search(None, "UNREAD")
    email_ids = response_data[0].split()

    if not email_ids:
        print("\nNo unread emails found in your inbox right now.")
        mail.logout()
        return

    print(f"Found {len(email_ids)} unread emails. Processing dynamic inference...\n")

    # Fetch the most recent 3 unread emails to avoid flooding the terminal
    for e_id in email_ids[-3:]:
        # Fetch the email body data
        status, data = mail.fetch(e_id, "(RFC822)")
        raw_email = data[0][1]

        # Parse raw bytes into a workable Email Object
        msg = email.message_from_bytes(raw_email)

        # Extract Meta information
        subject = clean_text_header(msg.get("Subject", "(No Subject)"))
        sender = clean_text_header(msg.get("From", "(Unknown Sender)"))

        # Extract the pure text Body from the email structure
        body = ""
        if msg.is_multipart():
            for part in msg.walk():
                content_type = part.get_content_type()
                content_disposition = str(part.get("Content-Disposition"))
                if content_type == "text/plain" and "attachment" not in content_disposition:
                    body = part.get_payload(decode=True).decode(
                        errors="ignore"
                    )
                    break
        else:
            body = msg.get_payload(decode=True).decode(errors="ignore")

        # Combine subject and body to give the model full context
        full_email_text = f"{subject} {body}".strip()

        if not full_email_text:
            continue

        # Execute Naive Bayes prediction
        prediction = live_pipeline.predict([full_email_text])[0]
        probabilities = live_pipeline.predict_proba([full_email_text])[0]
        classes = live_pipeline.classes_

        # Format probabilities for clean reading
        prob_scores = {
            classes[j]: f"{probabilities[j]:.2%}" for j in range(len(classes))
        }

        # Output the live decision logs
        print("-" * 50)
        print(f"From   : {sender}")
        print(f"Subject: {subject}")
        print(f"AI Tag : **{prediction.upper()}**")
        print(f"Scores : {prob_scores}")

    # Gracefully disconnect
    mail.logout()


if __name__ == "__main__":
    fetch_and_test_live_emails()
```
