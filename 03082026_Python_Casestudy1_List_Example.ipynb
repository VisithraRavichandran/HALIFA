```python
# List of existing house data: [SquareFeet, Bedrooms, Price]
house_database = [
    [1200, 2, 150000],
    [1500, 3, 200000],
    [1800, 3, 240000],
    [2200, 4, 300000],
    [2500, 4, 340000]
]

def calculate_average_rates(data):
    total_sqft_rate = 0
    total_bedroom_rate = 0

    # Calculate price per square foot and per bedroom across the list
    for house in data:
        sqft = house[0]
        bedrooms = house[1]
        price = house[2]

        # We split the price weight: 70% based on size, 30% based on rooms
        total_sqft_rate += (price * 0.70) / sqft
        total_bedroom_rate += (price * 0.30) / bedrooms

    avg_sqft_rate = total_sqft_rate / len(data)
    avg_bedroom_rate = total_bedroom_rate / len(data)

    return avg_sqft_rate, avg_bedroom_rate

def predict_house_price(sqft, bedrooms, avg_sqft_rate, avg_bedroom_rate):
    # Predict price using calculated average rates
    predicted_price = (sqft * avg_sqft_rate) + (bedrooms * avg_bedroom_rate)
    return round(predicted_price, 2)

# Main program execution
print("--- House Price Predictor ---")

# Step 1: Calculate market rates from the list data
avg_sqft_rate, avg_bedroom_rate = calculate_average_rates(house_database)

try:
    # Step 2: Get user inputs
    user_sqft = float(input("Enter target house size (Square Feet): "))
    user_bedrooms = int(input("Enter number of bedrooms: "))

    if user_sqft <= 0 or user_bedrooms <= 0:
        print("Please enter valid numbers greater than zero.")
    else:
        # Step 3: Run the prediction
        estimated_price = predict_house_price(user_sqft, user_bedrooms, avg_sqft_rate, avg_bedroom_rate)

        print("\n--- Prediction Result ---")
        print(f"Estimated Price for a {user_sqft} sqft, {user_bedrooms}-bedroom house is: ${estimated_price:,}")

except ValueError:
    print("Invalid input. Please enter numbers only.")
```
